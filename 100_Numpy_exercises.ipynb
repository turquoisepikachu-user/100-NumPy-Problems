#### 22. Normalize a 5x5 random matrix (★☆☆)

````python
a = np.random.uniform(-1000.0, 1000.0, size=(5, 5))
min_a = np.min(a)
max_a = np.max(a)
diff_a = max_a - min_a
norm_a = (a - min_a) / diff_a
print(norm_a)

print()

x = np.random.uniform(-1000.0, 1000.0, size=(5, 5))
def normalize(x):
    return (x - x.mean(axis=0)) / x.std(axis=0)
print(normalize(x))
````

#### 24. Multiply a 5x3 matrix by a 3x2 matrix (real matrix product) (★☆☆)

````python
A = np.arange(1, 16).reshape(5, 3)
B = np.arange(16, 22).reshape(3, 2)
AB = []

for i in range(A.shape[0]):
    elem = []
    for j in range(B.shape[1]):
        summa = 0
        for k in range(A.shape[1]):
            summa = summa + A[i][k] * B[k][j]
        elem.append(summa)
    AB.append(elem)
AB = np.array(AB)
print(AB)

print()

C = np.arange(-1, -16, -1).reshape(5, 3)
D = np.arange(-16, -22, -1).reshape(3, 2)
CD = np.dot(C, D)
print(CD)
````
